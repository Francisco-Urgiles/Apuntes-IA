# Conceptos y alcance de la IA
---
La IA se definió según 2 dimensiones principales:

- La fidelidad al desempeño humano
  > Copiar nuestros errores y sesgos

- La fidelidad a la racionalidad
  > Hacer lo correcto, tomar la mejor decision logica para lograr un objetivo

# **El TEST de Turing**
---

### ¿Qué es?
> Propuesto en 1950 evalua la capacidad de una máquina de mostrar un comportamineto inteligente e indintingible del de un ser humano

### ¿Qué buscaba?

> Este test buscaba fidelidad humana pero enfocado solo en el comportamiento, sin importar si tenia o no conciencia

Un agente Racional no intenta imitar la lógica humana sino que intenta dominar decisiones optimas, nace de fundamentos computacionales cruzadas con matemáticas puras y rigurosas con el cálculo de probabilidades para manejar la incertidumbre.

### Diferencias con matemáticas puras

Un modelo matemático es pasivo, analiza datos y arroja resultados, la IA toma esos datos y actua sobre su entorno de forma autónoma.

### El límite de la IA

Un algoritmo puede resultar muy util para tareas especializadas pero fuera de ellas falla.
La IA, por otro lado puede tener la flexibilidad cognitiva de un humano funcional, la capacidad de aprendizaje por transferencia ayuda a esta flexibilidad.

### La IA Simbolica
Esta resulta muy deficiente porque es suceptible a fallas, funciona en ambientes controlados y sigiendo reglas muy estrictas y poco flexibles

### Machine Learning
Es el proceso de aprendizaje computacional independiente usando redes neuronales artificiales, usando un entrenamiento basado en la identificación matemática de patrones, usando un algoritmo de retropropagación, esto resulta más tolerante a errores del entorno




# **Historia**
---
## **1950**

Allan Turing publica un documento llamado "Maquinaria Computacional e inteligencia" esto reformula la pregunta sobre la conciencia en las maquinas, propuso pruebas pragmáticas (El test de Turing).

Tambien exhibe el problema de la imposibilidad de codificar todo el conocimiento humano a mano, propuso como objetivo que una maquina "piense" con un cerebro con una arquitectura capaz de aprender

## **1956**
Un investigador llamado John McCarthy, junto con un grupo de cientificos, John McCarthy utilizó por 1ra vez el termino de Inteligencia Artificial, en este proyecto se propuso que la inteligencia era algo codificable en la inteligencia humana, estos pioneros tenían una fe ciega que resultó algo contraproducente, pese a que tenían avances como el desarrollo del programa teorico y lógico este programa logró demostrar teoremas matemáticos muy complejos.

## Micro Mundos
Los programas desarrollados en estas épocas resultaban tan eficientes ya que operaban en circunstancias controladas, y bajo reglas estrictas y fijas, ignorando que el mundo real es muy caótico, este exceso de confianza provocó ignorar este caos

## Sistemas expertos
El auge de algunos avances hizo que varios desarrolladores intentaran codificar cada especialidad bajo arboles de decisión que aunque resultaba parcialmente eficiente también tenía un gran problema y era la baja tolerancia de error, ante reglas tan estrictas bajo una falla conocida como "Explosión Combinatoria" que resulta de variaciones imprevistas estos sistemas arrojaban resultados erróneos.
Las posibilidades en el mundo real resultan tan impredecibles y grandes que el árbol de decisiones fuese tan grande y complejo que resulta ineficiente.

## Arthur Samuel
En 1959 él acuñó el término Machine Learning, su ejemplificación y demostración de esto fue un programa capaz de jugar Damas, pero la innovación es que no programó las reglas, le "enseñó" a evaluar el tablero, a partir de eso la maquina aprendió estrategias, a partir de miles de juegos e intentos (iteraciones) contra sí mismo, con ajustes matemáticos dependiendo del resultado (Ganar o perder). Pese a ello aún era parte de un Micro Mundo para adaptar esto al mundo real, debía cambiar su paradigma binario (Bueno o malo)
