#Paradigmas
---
>Para determinar si el problema pertenece a Clasificación o Regresión debes analizar que tipo de variable es.

> * Si la variable de respuesta (o de salida) es nominal u ordinal (Cualitativa) el problema es de **Clasificación**

> * Si la variable de respuesta (o de salida) es continua o discreta (Cuantitativas) el problema es de **Regresión**

##Regresión
